```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("dataset_Facebook.csv", delimiter=';')
df
```

```python
# Drop missing values (optional but useful)
df = df.dropna()
```

```python
#box plot
plt.figure(figsize=(8,5))
sns.boxplot(y=df['Page total likes'])

plt.title("Box Plot of Page Total Likes")
plt.show()
```

```python
#HISTOGRAM
plt.figure(figsize=(8,5))
plt.hist(df['Lifetime Post Total Reach'], bins=20)

plt.title("Histogram of Post Reach")
plt.xlabel("Reach")
plt.ylabel("Frequency")
plt.show()
```

```python
#SINGLE LINE GRAPH
plt.figure(figsize=(10,5))
plt.plot(df['like'])

plt.title("Single Line Graph - Likes")
plt.xlabel("Index")
plt.ylabel("Likes")
plt.show()
```

```python
#MULTIPLE LINE GRAPH
plt.figure(figsize=(10,5))

plt.plot(df['like'], label='Likes')
plt.plot(df['comment'], label='Comments')
plt.plot(df['share'], label='Shares')

plt.title("Multiple Line Graph")
plt.xlabel("Index")
plt.ylabel("Count")
plt.legend()

plt.show()
```
