```python
import pandas as pd

df = pd.read_csv("dataset_Facebook.csv", sep=';')

#Transposing Data
df_transposed = df.T
print("Transposed Data:")
print(df_transposed.head())
```

```python
#Sort Data
df_sorted = df.sort_values(by='Total Interactions', ascending=False)

print("Sorted Data:")
print(df_sorted.head())
```

```python
#Create Data Subsets
# Example: only Photo posts
df_subset = df[df['Type'] == 'Photo']

print("Subset Data:")
print(df_subset.head())
```

```python
#Merge Data
# Create another dataset
df_subset2 = df[['Type', 'Page total likes']]

# Merge
df_merged = pd.merge(df_subset, df_subset2, on='Type')

print("Merged Data:")
print(df_merged.head())
```
