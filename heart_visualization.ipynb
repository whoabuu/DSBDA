```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heart.csv")

print(df.head())
print(df.columns)
```

```python
#BOX PLOT
plt.figure(figsize=(8,5))
sns.boxplot(y=df['chol'])

plt.title("Box Plot of Cholesterol")
plt.show()
```

```python
#HISTOGRAM
plt.figure(figsize=(8,5))
plt.hist(df['age'], bins=20)

plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#SINGLE LINE GRAPH
plt.figure(figsize=(10,5))
plt.plot(df['thalach'])

plt.title("Single Line Graph - Max Heart Rate (thalach)")
plt.xlabel("Index")
plt.ylabel("Heart Rate")
plt.show()
```

```python
#MULTIPLE LINE GRAPH
plt.figure(figsize=(10,5))

plt.plot(df['age'], label='Age')
plt.plot(df['chol'], label='Cholesterol')
plt.plot(df['thalach'], label='Max Heart Rate')

plt.title("Multiple Line Graph - Heart Data Comparison")
plt.xlabel("Index")
plt.ylabel("Values")
plt.legend()

plt.show()
```
