```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("AirQualityUCI.xlsx")

# View data
print(df.head())
print(df.columns)
```

```python
# Replace -200 with NaN
df.replace(-200, pd.NA, inplace=True)

# Drop missing values
df = df.dropna()
```

```python
#BOX PLOT
plt.figure(figsize=(8,5))
sns.boxplot(y=df['CO(GT)'])

plt.title("Box Plot of CO(GT)")
plt.show()
```

```python
#HISTOGRAM
plt.figure(figsize=(8,5))
plt.hist(df['NO2(GT)'], bins=20)

plt.title("Histogram of NO2(GT)")
plt.xlabel("NO2 Concentration")
plt.ylabel("Frequency")
plt.show()
```

```python
#SINGLE LINE GRAPH
plt.figure(figsize=(10,5))
plt.plot(df['CO(GT)'])

plt.title("Single Line Graph - CO(GT)")
plt.xlabel("Index")
plt.ylabel("CO Level")
plt.show()
```

```python
#MULTIPLE LINE GRAPH
plt.figure(figsize=(10,5))

plt.plot(df['CO(GT)'], label='CO')
plt.plot(df['NO2(GT)'], label='NO2')
plt.plot(df['C6H6(GT)'], label='Benzene')

plt.title("Multiple Line Graph - Air Pollutants")
plt.xlabel("Index")
plt.ylabel("Concentration")
plt.legend()

plt.show()
```
