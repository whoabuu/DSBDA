```python
import requests
import bs4
```

```python
request1 = requests.get("https://www.amazon.in/X14STORE-Creative-Engraved-Decoration-Birthday/dp/B0DP8YF7MK/?_encoding=UTF8&pd_rd_w=hmNl4&content-id=amzn1.sym.31b48b2d-9c5c-4998-a8bc-a35f263d0c8b&pf_rd_p=31b48b2d-9c5c-4998-a8bc-a35f263d0c8b&pf_rd_r=ZNN7M94X7QAFF0YCP5WP&pd_rd_wg=FwNqr&pd_rd_r=766aed34-6b28-479a-a418-b64afed3e614&ref_=pd_hp_d_btf_NAMBOTTLES&th=1")
```

```python
request1.content
```

```python
soup=bs4.BeautifulSoup(request1.text)
```

```python
print(soup)
```

```python
rating=soup.find('span',{'class':'a-size-small a-color-base'}).get_text();
print(rating)
```

```python
reviews=soup.find_all('div',{'class':'a-expander-content reviewText review-text-content a-expander-partial-collapse-content'});
for review in reviews:
  print(review.get_text()+"\n\n")
```

```python
individual_rating=soup.find_all('i',{'class':'a-icon a-icon-star a-star-5 review-rating'});
for individual_r in individual_rating:
  print(individual_r.get_text()+"\n")
```

```python
customer_name=soup.find_all('span',{'class':'a-profile-name'});
for cust in customer_name:
  print(cust.get_text()+"\n")
```

```python
from IPython.display import Image,display
```

```python
imgs=soup.find_all('img',class_='a-dynamic-image a-stretch-horizontal media-block-image-tag')
for img in imgs:
  img_url = img['src']
  print(img_url)
  display(Image(url=img_url))
```
